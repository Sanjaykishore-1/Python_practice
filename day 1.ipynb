```python
var1 = int(input("Enter any number: "))
var2 = int(input("Enter any number: "))


sum = np.sum([var1, var2])
diff = np.diff([var1, var2])
product= np.product([var1, var2])
quotient = np.quotient([var1, var2])

print("the sum of two numbers is:", sum,
      "the difference of two numbers is:", diff,
      "the product of two numbers is:", product,
      "the quotient of two numbers is:", quotient)
```

```python
user_input = int(input("Enter any whole number: "))

if user_input % 2 == 0:
    print(user_input,"The number is even.")
else:
    print(user_input,"The number is odd.")
```

```python
user_input = int(input("Enter any number: "))

if user_input > 0:
    print(user_input,"The number is positive.")
elif user_input == 0:
    print(user_input,"The number is zero.")
else:
    print(user_input,"The number is negative.")
```

```python
user_input_1 = int(input("Enter first number: "))
user_input_2 = int(input("Enter second number: "))
user_input_3 = int(input("Enter a third number: "))

if user_input_1 > user_input_2 and user_input_1 > user_input_3:
    print(user_input_1, "is the largest number.")
elif user_input_2 > user_input_1 and user_input_2 > user_input_3:
    print(user_input_2, "is the largest number.")
else:
    print(user_input_3, "is the largest number.")
```
