```python
skills = {"python", "sql", "excel", "power bi","java"}

user_input = input("Enter your skill want to remove: ").lower()

if user_input in skills:
    skills.discard(user_input)
else:
    print("The given skill not match with user skills")    

print("The user skills are :", skills)
```

```python
numbers = (12, 5, 8, 20, 3, 15)

smallest = numbers[0]
largest  = 0
total = 0

for i in numbers:
    total += i

    if i > largest:
        largest = i
    elif i < smallest:
        smallest = i
        
print("smallest:",smallest)
print("largest:",largest)
print("Total:",total)
```

```python
def above_average(*numbers):
    above_avg = []

    for i in numbers:
        total = sum(numbers)
        avg = total/len(numbers)
        if i > avg:
            above_avg.append(i)  


    print("total :", total ,"Average:", avg)
    print(above_avg)
    print("Number of above average :", len(above_avg))           
```

```python
above_average(4,5,6)
```
