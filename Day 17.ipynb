```python
import numpy as np
import pandas as pd

np.random.seed(42)

students_df = pd.DataFrame({
    "Student": ["Maya", "Liam", "Sofia", "Noah", "Ava", "Ethan", "Emma", "Lucas", "Olivia", "Arjun"],
    "Math": np.random.randint(70, 101, size=10),
    "Science": np.random.randint(80, 101, size=10),
    "English": np.random.randint(80, 101, size=10)
})

students_df
```

```python
students_df.head(5)
students_df.shape
students_df.dtypes
students_df[["Math","Science","English"]].mean()
students_df[["Math","Science","English"]].max()
students_df[["Math","Science","English"]].min()
```

```python
students_df["Average"] = students_df[["Math", "Science", "English"]].mean(axis=1).round(2)
students_df[students_df["Average"] >= 90]
students_df[["Math","Science","English","Average"]]
```

```python
top_students= (
    students_df[students_df["Average"] >= 90]
    [["Math","Science","English","Average"]].sort_values("Average",ascending = False)  
)
top_students
```
