```python
try:
    user_name = input("Enter your name: ")
    user_goal = input("Enter your learning goal: ")

    with open("daily_notes.txt", "a") as file:
        print("Name:", user_name, file=file)
        print("Learning Goal :", user_goal, file=file)

    with open("daily_notes.txt", "r") as file:
        print(file.read())

except OSError:
    print("An error occurred while trying to read or write to the file.")
```

```python
while True:
    try:
        user_tasks = input("Enter your tasks for the day (or 'done' to exit): ")
        if user_tasks.lower() == 'done':
            break

        with open("tasks.txt.txt", "a") as file:
            print("Tasks for the day :", user_tasks, file=file)

        with open("tasks.txt.txt", "r") as file:
            print(file.read())

    except OSError:
        print("An error occurred while trying to read or write to the file.")       
```

```python
try:
    with open("tasks.txt.txt", "r") as file:
        tasks = file.readlines()

    for task_number, task in enumerate(tasks, start=1):
        print(task_number, task.strip())

    print("Total tasks:", len(tasks))

except FileNotFoundError:
    print("File not found.")

except OSError:
    print("An error occurred while trying to read or write to the file.")
```
