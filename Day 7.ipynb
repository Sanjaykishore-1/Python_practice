```python
def count_long_words(sentence):
    word_count = 0
    words = sentence.split()
    
    for word in words:
        if len(word) > 5:
            word_count += 1
    
    return word_count

    print(sentence)
    print("long count words are: ", word_count)            
```

```python
count_long_words("Python programming is interesting yaarrr")
```
