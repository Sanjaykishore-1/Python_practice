```python
import pandas as pd 

student_data = {
    "Maya": {"Math": 88, "Science": 92, "English": 85},
    "Liam": {"Math": 76, "Science": 81, "English": 79},
    "Sofia": {"Math": 95, "Science": 89, "English": 93},
    "Noah": {"Math": 84, "Science": 87, "English": 82},
    "Ava": {"Math": 91, "Science": 94, "English": 90}
}

students_df = pd.DataFrame(student_data).T

students_df["Total"] = students_df.sum(axis=1)
students_df["Average"] = students_df[["Math", "Science", "English"]].mean(axis=1).round(2)
students_df["Result"] = "Needs Improvement"
students_df.loc[students_df["Average"] >= 80, "Result"] = "Pass"

students_df
```

```python
students_df.loc[students_df["Total"]>= 250, "Grade"] = "A+"
students_df.fillna("B+")
```

```python
students_df["Average"].sort_values(ascending=False)
```

```python
students_df[["Average", "Result"]].sort_values(by="Average", ascending=False)[0:3]

#sorted_df[0:3]
```

```python
students_df.groupby("Result")["Average"].mean()
```

```python
math_avg = print(students_df["Math"].mean())
science_avg = print(students_df["Science"].mean())
english_avg = print(students_df["English"].mean())

```

```python
students_df.groupby("Average")[["Math","Science","English"]].mean()
```

```python
# filename: subject-averages.py
# Code Generated by Sidekick is for learning and experimentation purposes only.

subject_averages = students_df[["Math", "Science", "English"]].mean()
subject_averages
highest_average = subject_averages.max()
highest_average
print(f"The highest average is {highest_average}")
```

```python
subject_averages = {
    "Math": math_avg,
    "Science": science_avg,
    "English": english_avg
}

highest_average_sub = subject_averages.idxmax()
highest_average = subject_averages.max()

print(f"{highest_average_sub} has the highest average: {highest_average:.2f}")
```

```python
highest_subject = max(subject_averages, key=subject_averages.get)
highest_average = subject_averages[highest_subject]
highest_average
```
