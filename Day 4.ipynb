```python
numbers = [22, 45, 67, 89, 12, 34]

total = 0

for i in numbers:
    total = total + i
    print(total)

avg = total /len(numbers)    
print("the average is:", avg)
```

```python
numbers = [22, 45, 67, 89, 12, 34]

smallest = numbers[0]
for number in numbers:
    if number < smallest:
        smallest = number

print("The smallest number is:", smallest)
```

```python
numbers = [22, 45, 62, 89, 12, 34]

even = 0
odd = 0

for i in numbers:
    if i % 2 == 0: even += 1
    else: odd += 1
print(even)
print(odd)
```

```python
import numpy as np

numbers = [2, 4, 2, 6, 4, 8, 6, 10]

for number in numbers:
    unique_list = np.unique(numbers)
print(unique_list)    
```

```python
number = [2, 4, 2, 6, 4, 8, 6, 10]
unique_numbers = []

for n in number:
    if n not in unique_numbers:
        unique_numbers.append(n)

print(unique_numbers)
```

```python
numbers = [3, 12, 7, 20, 5, 18, 9]
greater_numbers = []

for n in numbers:
    if n >= 10:
        greater_numbers.append(n)

print(greater_numbers)
```

```python
numbers = [3, 12, 7, 20, 5, 18, 9]
square_numbers = []

for n in numbers:
    square_numbers.append(n*n)

print(square_numbers)
```
