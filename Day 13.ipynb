```python
def read_tasks(file_path):
    try:
        with open(file_path, "r") as file:
            return [line.strip() for line in file if line.strip()]

    except FileNotFoundError:
        print("File not found.")
        return None

    except OSError:
        print("An error occurred while reading the file.")
        return None


def summarise_tasks(tasks):
    completed_tasks = []
    pending_tasks = []

    for task in tasks:
        if "completed" in task.lower():
            completed_tasks.append(task)
        else:
            pending_tasks.append(task)

    print("Total tasks:", len(tasks))
    print("Completed tasks:", len(completed_tasks))
    print("Pending tasks:", len(pending_tasks))

    print("\nCompleted task details:")
    for task_number, task in enumerate(completed_tasks, start=1):
        print(f"{task_number}. {task}")


tasks = read_tasks("tasks.txt")

if tasks is not None:
    summarise_tasks(tasks)
```
