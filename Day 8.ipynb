Loop through the subjects: Math, Science, and English.

Store the highest score for each subject.
Store the student who achieved that score.
Print the result in this format:

```python
students = {
    "Maya": {"math": 88, "science": 92, "english": 85},
    "Liam": {"math": 76, "science": 81, "english": 79},
    "Sofia": {"math": 95, "science": 89, "english": 93}
}

high_mark_Math    = 0
high_mark_Science = 0
high_mark_English = 0


for student, subjects in students.items():
    for subject, mark in subjects.items():
        if subject.lower() == "math":
            if mark > high_mark_Math:
                high_mark_Math = mark
            
        if subject.lower() == "science":
            if mark > high_mark_Science:
                high_mark_Science = marks
            
        if subject.lower() == "english":
            if mark > high_mark_English:
                high_mark_English = marks    
            

print("Math high score: ", student, high_mark_Math)
print("science high score: ", student, high_mark_Science)
print("English high score:", student, high_mark_English)
print(student, "High score on :", subjects)
```

```python
students = {
    "Maya": {"math": 88, "science": 92, "english": 85},
    "Liam": {"math": 76, "science": 81, "english": 79},
    "Sofia": {"math": 95, "science": 89, "english": 93}
}

highest_scores = {
    "math": 0,
    "science": 0,
    "english": 0
}

for student, subjects in students.items():
    for subject, mark in subjects.items():
        if mark > highest_scores[subject]:
            highest_scores[subject] = mark

print(highest_scores)                
```
