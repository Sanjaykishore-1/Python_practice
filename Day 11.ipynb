```python
try:
    first_value = float(input("Enter first number: "))
    second_value = float(input("Enter second number: "))
    result = first_value/second_value
    print("The divided value of inputs are :", result)

except ValueError:
    print("Invalid input. Please enter a valid number.")

except ZeroDivisionError:
    print("Enter the second number other than zero. Please enter a valid number.")        
```

```python
students = {
    "Maya": {"Math": 88, "Science": 92, "English": 85},
    "Liam": {"Math": 76, "Science": 81, "English": 79},
    "Sofia": {"Math": 95, "Science": 89, "English": 93}
}

highest_average = 0

for student in students:
    try: 
        mark = sum(students[student].values())
        average = mark/len(students[student])
        print(f"The {student}'s average is: {average}")

        if average > highest_average:
            highest_average = average

    except KeyError:
        print("Enter a valid student name and subject.")

print(f"The top student is {student} with the highest average of {highest_average}")  
```
