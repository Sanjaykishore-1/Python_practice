```python
import random

names = [
    "Maya", "Liam", "Sofia", "Noah", "Ava", "Ethan", "Emma", "Lucas",
    "Olivia", "Mason", "Isabella", "James", "Amelia", "Henry", "Mia"
]
subjects = ["Math", "Science", "English"]

students = {
    student: {
        subject: random.randint(0, 100)
        for subject in subjects
    }
    for student in random.sample(names, 10)
}
```

```python
import pandas as pd

students_score = pd.DataFrame(students)
print(students_score)


def calculate_average(students_score):
    averages = students_score.mean(axis=0).round(2)

    for student, average in averages.items():
        print(f"The average of {student} is {average}")

    return averages


student_averages = calculate_average(students_score)
```
