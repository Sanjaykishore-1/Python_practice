```python
def analyze_text(word):
    characters = len(word)
    vovels = 0
    for letter in word:
        if letter in "aeio,u":
            vovels += 1
        elif letter in " ":
            characters -=1

    print("Characters excluding spaces:", characters )
    print("number of Vowels: ", vovels)
```

```python
analyze_text("Python is fun")
```

```python
def count_vowels(word):
    vovels = 0
    for letter in word.lower():
        if letter in "aeiou":
            vovels += 1
    print("number of Vowels: ", vovels)
```

```python
count_vowels("Python programming")
```
