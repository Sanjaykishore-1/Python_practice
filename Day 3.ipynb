```python
user_input_1 = float(input("Enter first number: "))
user_operator = input("Enter an operator (+, -, *, /): ")
user_input_2 = float(input("Enter second number: "))


if user_operator == "+":
    result = user_input_1 + user_input_2
    print("the result:", result)

elif user_operator == "-":
    result = user_input_1 - user_input_2
    print("the result:", result)

elif user_operator == "*":
    result = user_input_1 * user_input_2
    print("the result:", result)

elif user_operator == "/":
    if user_input_2 == 0:
        print("Error: Division by zero is not allowed.")
    else:
        result = user_input_1 / user_input_2
        print("the result:", result)

else:
    print("Invalid operator.")
```

```python
import numpy as np

positive = 0
negative = 0
zero = 0

number_count = int(input("Enter the number of inputs you want to enter: "))

for i in range(number_count):
    user_input = int(input("Enter a number: "))

    if user_input > 0:
        positive += 1
    elif user_input < 0:
        negative += 1
    else:
        zero += 1

print("Number of positive numbers entered:", positive)
print("Number of negative numbers entered:", negative)
print("Number of zeros entered:", zero)
```
